```python
"""
标准的对话模型的调用过程
    invoke()的输入可以是多种类型，典型的类型有
        1.字符串类型
        2.消息列表
    输出类型是：BaseMessage的子类：AIMessage
"""
from langchain_openai import ChatOpenAI

# 1.获取对话模型
llm = ChatOpenAI(
    model_name="GLM-5.1",# 默认使用的是gpt3.5
    base_url="xxx",
    api_key="xxx",
)

# 2.调用对话模型
resp = llm.invoke("什么是langChain？")

# 3.处理响应数据

print(resp.content)

# AIMessage
print(type(resp))
```
