```python
from langchain_core.messages import HumanMessage, SystemMessage

"""
原生支持多轮对话：
    通过消息列表维护上下文，模型可基于完整对话历史生成回复
适用场景：对话系统（如客服机器人，长期交互的AI助手）
"""

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model_name="GLM-5.1",# 默认使用的是gpt3.5
    base_url="",
    api_key="",
)

sys_message = SystemMessage(content='我是一个人工智能的助手，我的名字叫小智')

human_message=HumanMessage(content='猫王是一只猫吗？')

message = [sys_message, human_message]

resp = llm.invoke(message)

print(resp.content)

resp = llm.invoke('你叫什么名字')

print(resp.content)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage


from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model_name="GLM-5.1",# 默认使用的是gpt3.5
    base_url="",
    api_key="",
)

sys_message = SystemMessage(content='我是一个人工智能的助手，我的名字叫小智')

human_message1=HumanMessage(content='猫王是一只猫吗？')
human_message2=HumanMessage(content='你叫什么名字？')

message = [sys_message, human_message1,human_message2]

resp = llm.invoke(message)

print(resp.content)
```
