```python
"""
基础构造：适用于手动创建小规模数组或复制已有数据

np.array()
np.copy()
"""

import numpy as np
list1 = [1,2,3]
a = np.array(list1) # 是一个全新的数组
print(a)
list1[0] = 4
print(a)

a = np.copy(list1) # 是一个全新的数组
print(a)
list1[0] = 10
print(a)
```

```python
"""
预定义形状填充：用于快速初始化固定形状的数组

np.zeros()
np.ones()
np.empty()
np.full()

"""
"全0"

a = np.zeros((3,4),dtype=float)
print("全0:",a)


"全1"
b = np.ones((3,4),dtype=float)
print("全1:",b)

"未初始化"
c = np.empty((5,4),dtype=float)
print("未初始化：",c)

"固定值"
d = np.full((3,10),3)
print("固定值:",d)


"xxxx_likes：复制其结构"
dc = np.ones_like(d)
print("ones_like:",dc)
dc = np.zeros_like(d)
print("zeros_like:",dc)
dc = np.empty_like(d)
print("empty_like:",dc)
dc = np.full_like(d,888)
print("full_like：",dc)
```

```python
"""

基于数值范围生成：
np.arange()
np.linspace()
np.logspace()

"""

# 等差数列
arr = np.arange(1,51,2) # start end step。以start开始，每隔step取一个数
print(arr)

# 等间隔数列
arr = np.linspace(0,10,2) # start end num。在start和end之间取num个数，这num个数之间等间隔，包含start和end，取一个数，就是start
print(arr)

# 对数间隔数列
# start end num base。真正的start:start^base  真正的end:end^base，在真正的start和end上，分出num个数，包含start和end。
# 也可以换一种理解：log以base为底 x 为体 = start，那么x = base^start。其实我们先将start->end 等间隔分好，例如：0 5 10。然后在取base^(0 | 5 | 10)既可，就可以得出 1 32 1024
arr = np.logspace(0,10,3,base=2)
print(arr)
```

```python
"""
特殊矩阵生成：

零矩阵：所有元素为0
单位矩阵：对角线上为1，其余为0
对角矩阵：只有对角线有非零值
对称矩阵：A = A^T

np.eye()
np.diag()
"""

# 单位矩阵
arr = np.eye(3)
print(arr)

# 对角矩阵
arr = np.diag([5,1,2,3])
print(arr)


# 转置
arr = np.array([[1],[2],[3]])
print(arr)
print(arr.T)
```

```python
"""
随机数组的生成
np.random.rand()
np.random.randn()
np.random.randint()
"""

# 生成0到1之间的随机浮点数（均匀分布）
arr = np.random.rand(2,3) # 在[0,1)范围生成
print(arr)

# 生成指定范围区间的随机整数
arr = np.random.randint(2,3,(3,4)) # 在[2,3)范围生成
print(arr)

# 生成随机数列（正态分布）
arr = np.random.randn(2,3) # 在(-∞, +∞)范围生成，实际约 99.7% 在 (-3, 3) 之间，符合正态分布
print(arr)
```

```python
# 随机种子，相当于map

# 相当于存了一个1000-结果
np.random.seed(1000)
print(np.random.randint(1,10,(3,3)))

print(np.random.randint(1,10,(3,3)))

# 直接拿1000的结果
np.random.seed(1000)
print(np.random.randint(1,10,(3,3)))
```
