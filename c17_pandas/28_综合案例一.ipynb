```python
"""
企鹅数据分析
"""
# 1.导入必要的库
import pandas as pd
import numpy as np
# 2.导入数据
df = pd.read_csv('penguins.csv')
df.head(5)
df.info()
```

```python
# 3.数据清洗
# 缺失值的检查
df.dropna(inplace=True)
```

```python
# 4.数据特征的构造
df['sex'] = df['sex'].astype('category')
df['bill_ration'] = df['bill_length_mm'] / df['bill_depth_mm']
df.head()
```

```python
# 5.数据分析
# 数据分箱-把体重分为三个等级
labels = ['低','中','高']
df['mass_level'] = pd.cut(df['body_mass_g'],bins=3,labels=labels)
print(df['mass_level'].value_counts())

# 按岛屿，性别分组分析
df.groupby(['sex']).agg({
    'body_mass_g':['mean','count'],
})

df.groupby(['island']).agg({
    'body_mass_g':['mean','count'],
})




df.groupby(['sex','island']).agg({
    'body_mass_g':['mean','count'],
})
```
