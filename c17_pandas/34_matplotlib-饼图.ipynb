```python
# 饼图

import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体

# 设置字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10,5))

things = ['学习','娱乐','运动','睡觉','其它']
times = [4,3,1,8,5]
colors = ['red','blue','white','pink','black']
# 绘制饼图
plt.pie(times,labels=things,
        autopct='%.2f%%', # 显示百分比
        startangle=90) # 调整初始画图的角度
# # 添加标题
plt.title('一天时间分布',color='red',fontsize=10)

# 显示图表
plt.show()
```

```python
# 环形图

import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体

# 设置字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10,5))

things = ['学习','娱乐','运动','睡觉','其它']
times = [4,3,1,8,5]
colors = ['red','blue','white','pink','black']
# 绘制饼图
plt.pie(times,labels=things,
        autopct='%.2f%%', # 显示百分比
        startangle=90,# 调整初始画图的角度
        wedgeprops={'width':0.7},# 设置圆环的宽度
        pctdistance=0.7# 设置百分比的位置
        )
# # 添加标题
plt.title('一天时间分布',color='red',fontsize=10)
plt.text(0,0,'总计：\n100%%',ha='center',va='center')
# 显示图表
plt.show()
```

```python
# 饼图

import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体

# 设置字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10,5))

things = ['学习','娱乐','运动','睡觉','其它']
times = [4,3,1,8,5]
colors = ['red','blue','white','pink','black']
explode = [0.1,0,0,0,0] # 设置突出块的位置
# 绘制饼图
plt.pie(times,labels=things,
        autopct='%.2f%%', # 显示百分比
        startangle=0,# 调整初始画图的角度
        explode=explode,
        shadow=True,
        )
# # 添加标题
plt.title('一天时间分布',color='red',fontsize=10)

# 显示图表
plt.show()
```
