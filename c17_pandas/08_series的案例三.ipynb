```python
"""
给定某股票连续10个交易日的收盘价series
计算每日收益率（当日收盘价/前日收盘价 - 1）
找出收益率最高和最低的日期
计算波动率（收益率的标准差）
"""

import pandas as pd
import numpy as np

price = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2023-01-01',periods=10))
print(price)
```

```python
p1 = price.pct_change()
print("收益率变化")
print(p1)
```

```python
print(p1.idxmax())
print(p1.idxmin())

print(p1.std())
```
