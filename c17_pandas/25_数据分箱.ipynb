```python
# 数据分箱 pd.cut(x,bins,labels)
import pandas as pd

df = pd.read_csv('employees.csv')
df.head(10)
```

```python
df1 = df.head(10)[['employee_id','salary']]
df1
```

```python
print(df1['salary'].head(10))
pd.cut(df1['salary'].head(10),bins=2)
```

```python
pd.cut(df1['salary'].head(10),bins=2).value_counts()
```

```python
df1['收入范围'] = pd.cut(df1['salary'].head(10),bins=[0,10000,20000,30000],labels=['低','中','高'])
```

```python
df1
```

```python
pd.qcut(df1['salary'],3).value_counts()
```

```python
# 睡眠数据
df = pd.read_csv('sleep.csv')
df1 = df.head(10)[['person_id','sleep_quality']]
df1
```

```python
# 数值->分箱->统计
df['睡眠质量'] = pd.cut(df['sleep_quality'],bins=3,labels=['差','中','优'])
print(df['睡眠质量'].value_counts())
# 字符串->类别->统计
df['gender'] = df['gender'].astype('category')
print(df['gender'].value_counts())

print(df['睡眠质量'].dtypes)
print(df['gender'].dtypes)
```

```python
# df.rename() 给列或索引改名字（不改数据位置）
# df.set_inex() 把列"提升"为行索引，让数据更好查
# df.reset_inex() 把行索引"降级"为普通列，回到默认状态

df = pd.DataFrame({
    'name':['jack','alice','tom','bob'],
    'age':[18,19,20,21],
    'gender':['male','female','male','female'],
})

df.set_index('name',inplace=True)
print(df)
print(df.loc['jack','age'])
df.reset_index('name',inplace=True)
print(df)
df.rename(columns={'age':'年龄'},index={0:4})
```
