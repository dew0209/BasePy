```python
"""
每小时销售数据分析
某商店每小时销售额series
按天重采样计算每日总销售额
计算每天营业时间（8：00~22：00）和非营业时间的销售额比例
找出销售额最高的3个小时
"""

import pandas as pd
import numpy as np

np.random.seed(42)
values = np.random.randint(0,100,24)
hourly_sales = pd.Series(values,index=pd.date_range('2025-01-01',periods=24,freq='h'))
print(hourly_sales)
```

```python
# 按天重采样计算每日总销售额
day = hourly_sales.resample('D').sum()
print(day)


# 计算每天营业时间（8：00~22：00）和非营业时间的销售额比例
# 方式一：
# business_hours_sales = hourly_sales[(hourly_sales.index.hour>=8) & (hourly_sales.index.hour<=22)].sum()
# print(business_hours_sales / (day - business_hours_sales))
# 方式二：
# business_hours_sales = hourly_sales[(hourly_sales.index.hour>=8) & (hourly_sales.index.hour<=22)]
# not_business = hourly_sales.drop(business_hours_sales.index)
# print(business_hours_sales.sum() / not_business.sum())
# 方式三：
mask = (hourly_sales.index.hour>=8) & (hourly_sales.index.hour<=22)
b = hourly_sales[mask]
not_b = hourly_sales[~mask]
print(b.sum() / not_b.sum())
```

```python
# 找出销售额最高的3个小时
print(hourly_sales.nlargest(3).index)
```
