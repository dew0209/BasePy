```python
import pandas as pd
data = {
    'ID':[1,2],
    'name':['alice','bob'],
    'Math':[90,85],
    'English':[88,92],
    'History':[95,89]
}

df = pd.DataFrame(data)
df.T
df
```

```python
# 宽表转换成长表
df2 = pd.melt(df,id_vars=['ID','name'],var_name='科目',value_name='分数')
df2.sort_values('name')
```

```python
# 长表转宽表
pd.pivot(df2,index=['ID','name'],columns='科目',values='分数')
```
