```python
import pandas as pd

d = pd.Timestamp('2015-02-28 10:22')
d1 = pd.Timestamp('2015-02-28 11:32')
print(d)
print(type(d))
print('年：',d.year)
print('月：',d.month)
print('日：',d.day)
print(d.hour,"---",d.minute,"---",d.second)
print('季度：',d.quarter)
print('是否月底：',d.is_month_end)

# 方法
print('星期几：',d.day_name())
print('转换为天：',d.to_period("D"))
print('转换为季度：',d.to_period("Q"))
print('转换为年度：',d.to_period("Y"))
print('转换为月度：',d.to_period("M"))
print('转换为周维度：',d.to_period("W"))
```

```python
# 将字符春转化为日期类型

a = pd.to_datetime('2015-02-28')
print(a)
print(type(a))
print(a.day_name())
```

```python
# dataFrame日期转换
df = pd.DataFrame({
    'sales':[100,200,300],
    'data':['20250601','20250602','20250603']
})
df['dateTime'] = pd.to_datetime(df['data'])
print(df['dateTime'].dt.day_name())
```

```python
# 时间间隔
d1 = pd.Timestamp('2013-01-15')
d2 = pd.Timestamp('2023-01-15')
d3 = d2 - d1
print(type(d3))
print(d3)
```
