```python
"""
head():查看前n行数据，默认5行
tail():查看后n行数据，默认5行
isin():判断元素是否包含在参数集合中
isna():判断是否为缺失值（如NaN或None）
sum():求和，自动忽略缺失值
mean():平均值
min():最小值
max():最大值
var():方差
std():标准差
median():中位数
mode():众数（可返回多个）
quantile(q):分位数，q取0~1之间
describe():常见统计信息（count，mean，st，min，25%，75%，max）
value_counts():每个唯一值的出现次数
count():非缺失值数量
duplicated():是否重复
drop_duplicates():去除重复项
sample():随机抽样
replace():替换值
sort_index():按索引排序
sort_values():按值排序
replace():替换值
nlargest():返回某列最大的n条数据
nsmallest():返回某列最小的n条数据

"""

import pandas as pd
import numpy as np

# 通过Series来创建
df = pd.DataFrame({"id":[1,2,3,4,5],
                   "name":['tom','jack','bob','alice','zoom'],
                   "age":[20,30,40,50,10]
                   },index=[f"学生{i}" for i in range(1,6)],columns=['id','age','name'])
print(df)
```

```python
print(df.isin(['jack','学生1',20])) # 查看元素是否包含在参数集合中
```

```python
print(df.isna())
```

```python
print(df.sum())
print(df['age'].sum())
print(df.age.max())
print(df.age.min())
print(df.age.mean())
print(df.age.median())
print(df.age.var())
print(df.age.std())
print(df.age.quantile(0.25))
```

```python
print(df.describe())
```

```python
print(df.count())
```

```python
df = pd.DataFrame({"id":[1,1,3,4,5],
                   "name":['tom','tom','bob','alice','zoom'],
                   "age":[20,20,40,50,10]
                   },index=[f"学生{i}" for i in range(1,6)],columns=['id','age','name'])

print(df.value_counts())
```

```python
print(df.drop_duplicates())
```

```python
print(df.duplicated(subset=['age']))
```

```python
print(df.sample(2))
```

```python
print(df.replace(20,1))
```

```python
print(df.cumsum())
print(df.cummax())
print(df.cummin())
```

```python
print(df.sort_index(ascending=False))
```

```python
print(df.sort_values(by=['age'],ascending=[False]))
```

```python
print(df.nlargest(2,columns=['age']))
print(df.nsmallest(2,columns=['age']))
```
