```python
import pandas as pd

# series的创建
s = pd.Series([10,2,3,1,4])
print(s)
# 自定义索引
s = pd.Series([10,2,3,1,4],index=['a','b','c','d','e'])
print(s)
# 定义name
s = pd.Series([1,2,3,4,5,6],name='测试')
print(s)
# 通过字典来创建
s = pd.Series({"a":1,"b":2,"c":3,"d":4,"e":5})
print(s)
# 通过存在的series创建
s = pd.Series(s,index=['a','e'])
print(s)
```
