```python
import pandas as pd
import numpy as np

data = {
    'name':['alice','alice','bob','alice','jack','bob'],
    'age':[26,25,30,25,35,30],
    'city':['NY','NY','LA','NY','SF','LA']
}
df = pd.DataFrame(data)
df
```

```python
df.duplicated() # 一整条记录都是一样的，标记为重复，返回True
print(df.drop_duplicates(subset=['name']) ) # 根据指定的列去重
print(df.drop_duplicates(subset=['name'],keep='last'))  # 根据指定的列去重 保留最新的
```
