```python
import pandas as pd

"""
给定某城市一周每天的最高温度Series，完成以下任务：
找出温度超过30度的天数
计算平均温度
将温度从高到低排序
找出温度变化最大的两天
"""
import pandas as pd
import numpy as np

temperatures = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])

print('超过30度的天数：',temperatures[temperatures > 30].count())
print('平均温度：',temperatures.mean())
t2 = temperatures.sort_values(ascending=False)
print('从高到低的排序：',t2)

# 计算series数组变化值
t3 = temperatures.diff().abs()
print(t3.sort_values(ascending=False).keys()[0:2].tolist())
```
