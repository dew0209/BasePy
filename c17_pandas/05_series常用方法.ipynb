```txt
方法                              说明
head()                           查看前n行数据，默认5行
tail()                           查看后n行数据，默认5行
isin()                           判断元素是否包含在参数集合中
isna()                           判断是否为缺失值
sum()                            求和，自动忽略缺失值
mean()                           平均值
min()                            最小值
max()                            最大值
var()                            方差
std()                            标准差
median()                         中位数
mode()                           众数（可返回多个）
quantile(q)                      分位数，q取0~1之间
describe()                       常见统计信息（count,mean,std,min,25%,50%,75%,max）
value_counts()                   每个唯一值的出现次数
count()                          非缺失值数量
nunique()                        唯一值个数（去重）
unique()                         获取去重后的值数组
drop_duplicates()                去除重复项
sample()                         随机抽样
sort_index()                     按索引排序
sort_value()                     按值排序
replace()                        替换值
keys()                           返回Series的索引对象
```

````python
import pandas as pd
import numpy as np

s = pd.Series([10,2,np.nan,None,3,4,5],index=['a','b','c','d','e','f','g'],name='data')
print(s)
````

````python
print(s.head())
print(s.tail())
````

````python
s.describe()
````

````python
# 获取元素个数 忽略缺失值
print(s.count())
````

````python
# 获取索引
print(s.keys())
print(s.index)
````

````python
print(s.isna()) # 检查Series中每一个元素是否是缺失值
````

````python
s.isin([10,4]) # 检查每个元素是否在参数集合中
````

````python
print(s.mean()) # 平均值
print(s.sum())  # 综合
print(s.std())  # 标准差
print(s.var())  # 方差
print(s.min())  # 最小值
print(s.max())  # 最大值
print(s.median()) # 中位数
print(s.quantile(0.40)) # 分位
print(s.mode()) #
````

````python
print(s.value_counts()) # 统计每个元素的出现数量
````

````python
s = pd.Series([1,2,1,4,6,5,5])
print(s.drop_duplicates()) # 返回的是Series
print(s.unique()) # 返回的是一个列表
print(s.nunique()) # 去重后的个数
````

````python
s = pd.Series([10,2,np.nan,None,3,4,5],index=['a','b','c','d','e','f','g'],name='data')

# 排序
print(s.sort_index())
print(s.sort_values())
````
