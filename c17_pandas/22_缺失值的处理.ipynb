```python
import pandas as pd
import numpy as np

s = pd.Series([1,2,pd.NA,np.nan,None])
print(s)

# 查看是否有缺失值
print(s.isna())
print(s.isnull())


df = pd.DataFrame([[1,pd.NA,2],[2,3,4],[None,4,6]],columns=['列1','列2','列3'])

print(df.isna())
print('-----')

print(df.isnull())
print('-----')
print(df.isna().sum())
print(s.isna().sum())
```

```python
# 剔除缺失值
print(s.dropna())
print('*' * 30)
print(df.dropna())
print('*' * 30)
df = pd.DataFrame([[np.nan,pd.NA,None],[2,3,4],[None,4,6]],columns=['列1','列2','列3'])
print(df.dropna(how='all')) # 如果所有值都是缺失值，删除这一行
print('*' * 30)
print(df.dropna(thresh=3)) # 如果至少有n个值不是缺失值，就保留
print('*' * 30)
df = pd.DataFrame([[np.nan,1,None],[2,3,4],[None,4,6]],columns=['列1','列2','列3'])
print(df.dropna(axis=1))
print('*' * 30)
print(df.dropna(subset=['列1'])) # 如果某列有缺失值，则删除这一行
```

```python
# 填充缺失值

df = pd.read_csv('weather_withna.csv')
print(df.fillna({'temp_max':20,'wind':2.5}).tail()) # 使用字典来填充
print(df.fillna(df[['temp_max','wind']].mean()).tail()) # 使用统计值来填充
print(df.ffill().tail()) # 用前面的相邻值填充
print(df.bfill().tail()) # 用后面的相邻值填充
```
