```python
"""
某产品过去12个月的销售量series
计算季度平均销量（每3个月为一个季度）
找出销量最高的月份
计算月环比增长率
找出连续增长超过2个月的月份

"""
```

```python
import pandas as pd
import numpy as np

sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],
                  index=pd.date_range(start='2022-01-01',periods=12,freq='MS'))
print(sales)
```

```python
# 季度的平均销量
print(sales.resample('QS').mean()) # 重新采样

# 效率最高的月份
print(sales.idxmax())

sales.pct_change()
```

```python
# 计算月环比增长率
a = sales.pct_change();
a = a > 0
a[a.rolling(3).sum()==3].keys().tolist()
```
