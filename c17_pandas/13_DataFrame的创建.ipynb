```python
# dataFrame的创建

import pandas as pd
import numpy as np

# 通过Series来创建
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([3,1,34,5,6])
df = pd.DataFrame({"s1":s1,"s2":s2})
print(df)
print(type(df))
print(df['s1'])
```

```python
# 通过字典来创建
df = pd.DataFrame({"id":[1,2,3,4,5],
                   "name":['tom','jack','bob','alice','zoom'],
                   "age":[20,30,40,50,10]
                   },index=[f"学生{i}" for i in range(1,6)],columns=['id','age','name'])
print(df)
```
