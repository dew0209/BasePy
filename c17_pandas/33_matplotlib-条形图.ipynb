```python
# 柱状图

import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体

# 设置字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10,5))

subjects = ['语文','数学','英语','科学']
scores = [85,92,78,88]

# 绘制柱状图
plt.bar(subjects,scores,label='小红',width=0.4,color='orange')
# 添加标题
plt.title('小红2026年成绩分布',color='red',fontsize=10)
# 添加坐标轴的标签
plt.xlabel('科目',fontsize=10)
plt.ylabel('成绩',fontsize=10)
# 添加图例
plt.legend(loc='upper right',fontsize=10)
# 添加网格线
plt.grid(axis='y',alpha=0.1,color='blue',linestyle='--')

# 自定义刻度字体大小
plt.xticks(rotation=60,fontsize=5)
plt.yticks(fontsize=10)

# 设置y轴范围
plt.ylim(0,100)

# 在每个数据点上显示数值

for x,y in zip(subjects,scores):
    plt.text(x,y + 1,str(y),ha='center',va='bottom',fontsize=10)


# 显示图表
plt.show()
```

```python
# 条形图

import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体

# 设置字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10,5))

countries = ['Unite States','China','Japan','India']
gdp = [111,100,99,58]

# 绘制柱状图
plt.barh(countries,gdp,color='orange')
# # 添加标题
plt.title('2026年GDP分析',color='red',fontsize=10)

# 显示图表
plt.show()
```
