```txt
属性                          说明
index                           Series的索引对象
values                          Series的值
dtype或dtypes                   Series的元素类型
shape                           Series的形状
ndim                            Series的维度
size                            Series的元素个数
name                            Series的名称
loc[]                           显式索引，按标签索引或切片
iloc[]                          隐式索引，按位置索引或切片
at[]                            使用标签访问单个元素
iat[]                           使用位置访问单个元素
```

````python
import pandas as pd

s = pd.Series([1,2],index=['a','b'])
print(s.index)
print(s.values,s.ndim,s.size)
s.name = 'test'
print(s.dtype,s.name)
print(s.shape)
print("显示索引:",s.loc['a':'a']) # 显示索引
print("隐式索引:",s.iloc[0:1])  # 隐式索引
print(s.at['a'])
print(s.iat[1])
````
